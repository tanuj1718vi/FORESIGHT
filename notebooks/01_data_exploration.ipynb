# 🔮 FORESIGHT — 01: Exploratory Data Analysis & Demand Intelligence

Welcome to the **FORESIGHT** exploratory intelligence notebook. This notebook audits demand patterns, seasonality, price elasticity, ABC/XYZ product segmentation, and inventory turnover dynamics across 50 SKUs and 5 fulfillment centers.

```python
import pandas as pd
import numpy as np
from foresight.data.loader import load_processed_sales
from foresight.data.eda import EDAEngine
from foresight.data.quality import DataQualityEngine

# 1. Load Processed Dataset
df = load_processed_sales()
print(f"Loaded dataset: {len(df):,} rows across {len(df.columns)} columns")
df.head()
```

## 2. Automated Data Quality Verification
Run automated checks across all 11 integrity and schema constraints.

```python
quality_engine = DataQualityEngine()
quality_report = quality_engine.evaluate(df, dataset_name="sales_processed")
print(f"Quality Score: {quality_report.quality_score:.1f}% | Status: {quality_report.overall_status.value}")
for check in quality_report.checks:
    print(f"  - [{check.status.value}] {check.check_name}: {check.message}")
```

## 3. Comprehensive Statistical Demand Profiling
Analyze overall demand volume, dispersion, promotional lift, and multi-year market trend.

```python
eda_engine = EDAEngine()
report = eda_engine.generate_full_report(df)
dp = report.demand_profile

print(f"Total Sales Units: {dp.total_sales_units:,}")
print(f"Total Gross Revenue: ${dp.total_revenue:,.2f}")
print(f"Mean Daily Demand: {dp.mean_daily_sales:.2f} units/day")
print(f"Promotional Lift Factor: {dp.promotional_demand_lift:.2f}x")
print(f"Annual Market Trend: {dp.annualized_trend_growth_pct:+.2f}% / year")
```

## 4. Interactive Demand Trend & Rolling Averages (Plotly)

```python
fig_trend = eda_engine.plot_demand_trend_and_seasonality(df)
fig_trend.show()
```

## 5. Day-of-Week Seasonality by Category

```python
fig_season = eda_engine.plot_day_of_week_seasonality(df)
fig_season.show()
```

## 6. ABC / XYZ Portfolio Segmentation Matrix
Segmenting catalog SKUs by cumulative revenue contribution (ABC) and volatility coefficient of variation (XYZ).

```python
seg = report.segmentation
print(f"Total SKUs Analyzed: {seg.total_skus}")
print("Segment Breakdown:")
for s, count in sorted(seg.segment_counts.items()):
    print(f"  Segment {s}: {count} SKUs ({count / seg.total_skus * 100:.1f}%)")

fig_matrix = eda_engine.plot_abc_xyz_matrix(df)
fig_matrix.show()
```

## 7. Inventory Health, Stockouts & Turnover Dynamics

```python
ih = report.inventory_health
print(f"Working Inventory Value: ${ih.average_inventory_value:,.2f}")
print(f"Annualized Inventory Turnover: {ih.annualized_inventory_turnover:.2f}x / year")
print(f"Average Days of Supply: {ih.overall_days_of_supply:.1f} days")
print(f"Stockout Rate: {ih.stockout_rate_pct:.2f}%")
print(f"Total Backordered Units: {ih.total_backordered_units:,}")

top_sku = seg.top_revenue_skus[0]
fig_inv = eda_engine.plot_sku_inventory_profile(df, sku_id=top_sku, store_id="STORE-001")
fig_inv.show()
```
