# 🔮 FORESIGHT — 06: Rolling-Origin Model Evaluation & Leaderboard

This notebook demonstrates enterprise-grade temporal cross-validation:
- **Rolling-Origin Walk-Forward CV** without lookahead leakage.
- Metric benchmarking across folds (WAPE, RMSE, MAE, sMAPE, R²).
- Residual diagnostics and error distribution analysis.

```python
import pandas as pd
import numpy as np
from foresight.evaluation.benchmark import ModelBenchmarkRunner

df = pd.read_parquet('data/processed/features_engineered.parquet')
print(f"Loaded feature matrix: {len(df):,} observations")
```

## 2. Execute Rolling Origin Cross-Validation (3 Folds, 30-Day Horizons)

```python
runner = ModelBenchmarkRunner(n_splits=3, horizon_days=30, primary_metric="wape")
report = runner.run(df)

print(f"🏆 Champion Model: {report.champion_model_name} (Mean WAPE: {report.champion_mean_wape:.4f})")
print("\nLeaderboard Markdown Report:")
print(report.to_markdown())
```
